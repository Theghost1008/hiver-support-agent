```python
import pandas as pd
from pathlib import Path

DATA_PATH = Path('../data/raw/twcs_subsample.csv')
df = pd.read_csv(DATA_PATH)

df.shape
```

```python
df.head()
```

```python
brand_counts = df[df["inbound"]==False]["author_id"].value_counts()
brand_counts.head(20) 
```

```python
is_apple_reply = df["author_id"] == "AppleSupport"
is_customer_to_apple = (df["inbound"]==True) & (df["text"].str.contains("@AppleSupport",case=False, na=False))

apple_df = df[is_apple_reply | is_customer_to_apple].copy()
apple_df.shape
```

```python
customer_msgs = apple_df[apple_df["inbound"]==True]
brand_replies = apple_df[apple_df["inbound"]==False]

pairs = customer_msgs.merge(
    brand_replies,
    left_on="tweet_id",
    right_on="in_response_to_tweet_id",
    suffixes=("_customer","_apple")
)
pairs[["text_customer","text_apple"]].head(10)
```
